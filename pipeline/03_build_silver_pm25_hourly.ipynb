```python
spark.sql("""
CREATE OR REPLACE TABLE workspace.delhi_air.silver_pm25_hourly AS
SELECT sensor_id, location_id, location_name, provider,
       date_trunc('hour', period_start_utc + INTERVAL 330 MINUTES) AS hour_ist,
       round(avg(pm25_ugm3), 2) AS pm25_avg,
       count(*) AS readings,
       CASE WHEN max(interval_text) = '00:15:00' THEN 4 ELSE 1 END AS expected_readings
FROM workspace.delhi_air.silver_pm25
GROUP BY 1, 2, 3, 4, 5
""")
print("done")
```
